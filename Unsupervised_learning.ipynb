```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import pandas as pd
import numpy as np
import random
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x, y_true = make_blobs(n_samples=500,cluster_std=0.5, random_state=0)
df = pd.DataFrame(x, columns = ["feature1", "feature2"])
```

```python
df.head()
```

```python
le = StandardScaler()
```

```python

x_Scaled = le.fit_transform(df)
x_Scaled
```

elbow method

```python
from sklearn.cluster import KMeans
```

```python
inertia = []
x_range = range(1, 11)
for i in x_range:
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(x_Scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(x_range, inertia, marker = "o")
```

```python
kmeans_final = KMeans(n_clusters = 3)
cluster = kmeans_final.fit_predict(x_Scaled)
```

```python
df["cluster"] = cluster
```

```python
df.columns
```

```python
sns.scatterplot(
    x="feature1", y="feature2", hue="cluster", data=df, palette="viridis"
)
```

```python
from sklearn.datasets import make_moons
```

```python
x, y = make_moons(n_samples=1000, noise=0.5, random_state=42)
```

```python
df = pd.DataFrame(x, columns = ["feature1", "feature2"])
```

```python
le = StandardScaler()
```

```python
x_Scaled = le.fit_transform(df)
```

```python
from sklearn.cluster import DBSCAN
```

```python
clustering = DBSCAN(eps=0.3, min_samples=2).fit(x_Scaled)
cluster = clustering.labels_
```

```python
cluster
```

```python
df["cluster"] = cluster
```

```python
sns.scatterplot(
    x="feature1", y="feature2", hue="cluster", data=df, palette="viridis"
)
```
