```python
import string
def remove_punc(text):
  for punc in string.punctuation:
    text = text.replace(punc,"")
  return text
```

```python
df["text"] = df["text"].apply(remove_punc)
df["text"]
```

```python
def remove_num(text):
  for i in text:
    if i.isdigit():
      text = text.replace(i,"")
  return text
```
