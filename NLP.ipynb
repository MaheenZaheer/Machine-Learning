```python
df["text"] = df["text"].apply(lambda x :x.lower())
```

```python
import string
def remove_punc(text):
  for punc in string.punctuation:
    text = text.replace(punc,"")
  return text
```

```python
df["text"] = df["text"].apply(remove_punc)
df["text"]
```

```python
def remove_num(text):
  for i in text:
    if i.isdigit():
      text = text.replace(i,"")
  return text
```

```python
df["text"] = df["text"].apply(remove_num)
df["text"]
#we do not have emjois here but if we have then simply use isascii since ascii is for the txt and numbers but the unicode is for everything so we can use above for loop for that purpose
```

```python
#remove stopword it is necessary in nlp but not in deep learning
from nltk.corpus import stopwords
import nltk
nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize
nltk.download("punkt_tab")
```

```python
stop_words = set(stopwords.words("english"))
```

```python
# def remove_stopwords(text):
#   word_tokens = word_tokenize(text)
#   filtered_text = [word for word in word_tokens if word not in stop_words]
#   return " ".join(filtered_text)
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    words = []

    for word in tokens:
        if word not in stop_words:
            words.append(word)

    return " ".join(words)
```

```python
df["text"] = df["text"].apply(remove_stopwords)
df["text"]
```
