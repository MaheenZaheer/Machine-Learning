```python
import string
def remove_punc(text):
  for punc in string.punctuation:
    text = text.replace(punc,"")
  return text
```

```python
df["text"] = df["text"].apply(remove_punc)
df["text"]
```

```python
def remove_num(text):
  for i in text:
    if i.isdigit():
      text = text.replace(i,"")
  return text
```

```python
df["text"] = df["text"].apply(remove_num)
df["text"]
#we do not have emjois here but if we have then simply use isascii since ascii is for the txt and numbers but the unicode is for everything so we can use above for loop for that purpose
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    words = []

    for word in tokens:
        if word not in stop_words:
            words.append(word)

    return " ".join(words)
```

```python
# we have techniques for vectorization like bag of words one hot encoding (failed search it on google why ) tf-idf
# currently we gonna use bag of words it works on frequency for this first of all we goona convert our txt in vocab
# after thta from that vocab we calculate each word how many it is apperaed in sentaence or documnet samilarly we do this for all coupus
```

```python
# tf-idf (term freq and inverse document freq)
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df["text"], df["emotion"], test_size=0.2, random_state=42)
```

```python
# now lets convert our x-test and x-train in bag of vectorizer
vectorizer = CountVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Fit MultinomialNB directly on sparse matrices
model = MultinomialNB()
model.fit(X_train_vectorized, y_train)

# Predict without .toarray()
y_pred = model.predict(X_test_vectorized)

# Evaluate accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc}")
```

```python
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# Predict without .toarray()
y_pred = model.predict(X_test_tfidf)

# Evaluate accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc}")
```
